```python
# ---- Rendering defaults (tuned for legibility) ----
NODE_COLOR = "#b4e8f3"
EDGE_COLOR = "#d3d3d3"
EDGE_WIDTH = 1.6
ARROWSIZE = 24
FONT_SIZE = 14
DEFAULT_MAX_CHARS = 14

EXPORT_GRAPHML = "final_dag.graphml"
EXPORT_GML = "final_dag.gml"
EXPORT_PNG = "graph_visualization.png"
```

```python
def _safe_strip(x: object) -> str:
    if pd.isna(x):
        return ""
    return str(x).strip()

def wrap_label(text: str, max_chars: int = DEFAULT_MAX_CHARS) -> str:
    """
    Wrap a label onto multiple lines without splitting words.
    Breaks occur at underscores or spaces.
    """
    import re

    text = (text or "").replace("_", " ")
    words = re.split(r"(_|\s)", text)

    lines = []
    current = ""
    for w in words:
        projected = len(current) + len(w)
        if projected > max_chars and current:
            lines.append(current)
            current = w if not re.match(r"(_|\s)", w) else ""
        else:
            current += w

    if current:
        lines.append(current)

    return "\n".join(lines)


def load_nodes(nodes_path: str) -> pd.DataFrame:
    df = pd.read_csv(nodes_path, dtype=str).fillna("")
    required = [
        "node_id",
        "node_name",
        "definition_quote",
        "parent_node",
        "explicitness",
        "internal_inconsistency",
        "notes",
    ]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"nodes file is missing columns: {missing}")
    df["node_id"] = df["node_id"].map(_safe_strip)
    df["node_name"] = df["node_name"].map(_safe_strip)
    return df


def load_edges(edges_path: str) -> pd.DataFrame:
    df = pd.read_csv(edges_path, dtype=str).fillna("")
    required = [
        "source",
        "target",
        "mechanism",
        "temporal_structure",
        "context",
        "necessity_sufficiency",
        "supporting_quote",
    ]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"edges file is missing columns: {missing}")
    df["source"] = df["source"].map(_safe_strip)
    df["target"] = df["target"].map(_safe_strip)
    return df


def build_graph(nodes_df: pd.DataFrame, edges_df: pd.DataFrame) -> nx.DiGraph:
    # Map node_id -> node_name (prefer node_name if present)
    id_to_name = {}
    for _, row in nodes_df.iterrows():
        nid = row["node_id"]
        name = row["node_name"] if row["node_name"] else nid
        if not nid:
            continue
        id_to_name[nid] = name

    G = nx.DiGraph()

    # Add nodes (use node_id as stable key, store display name as attribute)
    for _, row in nodes_df.iterrows():
        nid = row["node_id"]
        if not nid:
            continue
        label = _safe_strip(row["node_name"]) or nid
        G.add_node(
            nid,
            label=label,
            definition_quote=_safe_strip(row["definition_quote"]),
            parent_node=_safe_strip(row["parent_node"]),
            explicitness=_safe_strip(row["explicitness"]),
            internal_inconsistency=_safe_strip(row["internal_inconsistency"]),
            notes=_safe_strip(row["notes"]),
        )

    # Add edges
    for _, row in edges_df.iterrows():
        src = row["source"]
        dst = row["target"]
        if not src or not dst:
            continue

        # If edges refer to node_name instead of node_id, attempt to map by label.
        src_id = src if src in G.nodes else None
        dst_id = dst if dst in G.nodes else None

        if src_id is None:
            matches = [nid for nid, data in G.nodes(data=True) if data.get("label", "") == src]
            src_id = matches[0] if matches else src
            if src_id not in G.nodes:
                G.add_node(src_id, label=src_id)

        if dst_id is None:
            matches = [nid for nid, data in G.nodes(data=True) if data.get("label", "") == dst]
            dst_id = matches[0] if matches else dst
            if dst_id not in G.nodes:
                G.add_node(dst_id, label=dst_id)

        G.add_edge(
            src_id,
            dst_id,
            mechanism=_safe_strip(row["mechanism"]),
            temporal_structure=_safe_strip(row["temporal_structure"]),
            context=_safe_strip(row["context"]),
            necessity_sufficiency=_safe_strip(row["necessity_sufficiency"]),
            supporting_quote=_safe_strip(row["supporting_quote"]),
        )

    return G


def warn_if_cycles(G: nx.DiGraph) -> None:
    if nx.is_directed_acyclic_graph(G):
        print("OK: Graph is a DAG (acyclic).")
        return

    print("WARNING: Graph contains cycles (not a DAG).", file=sys.stderr)
    try:
        cycles = list(nx.simple_cycles(G))
        print(f"Found {len(cycles)} cycle(s). Showing up to 5:", file=sys.stderr)
        for cyc in cycles[:5]:
            labels = [G.nodes[n].get("label", n) for n in cyc]
            print("  - " + " -> ".join(labels) + " -> " + labels[0], file=sys.stderr)
    except Exception as e:
        print(f"Could not enumerate cycles: {e}", file=sys.stderr)

def hierarchical_layout(
    G: nx.DiGraph,
    direction: str = "LR",
    node_size: float = 3800,
    nodesep: float = 0.9,
    ranksep: float = 1.5,
):
    """
    Graphviz dot layout with explicit spacing + direction.
    Requires pygraphviz.
    """
    from networkx.drawing.nx_agraph import graphviz_layout

    # Encourage dot to allocate enough space per node
    # This is a heuristic; Graphviz uses inches for width/height.
    node_dim = (node_size ** 0.5) / 180.0
    for n in G.nodes:
        G.nodes[n]["width"] = str(node_dim)
        G.nodes[n]["height"] = str(node_dim)
        G.nodes[n]["shape"] = "box"

    args = f"-Grankdir={direction} -Gnodesep={nodesep} -Granksep={ranksep}"
    return graphviz_layout(G, prog="dot", args=args)


def _layout(G: nx.DiGraph, direction: str = "LR", node_size: float = 3800):
    # Prefer Graphviz dot layout; fallback to spring if unavailable.
    try:
        return hierarchical_layout(G, direction=direction, node_size=node_size)
    except Exception:
        return nx.spring_layout(G, seed=7)


def _auto_figsize(n_nodes: int, base: Tuple[float, float] = (22, 16)) -> Tuple[float, float]:
    """
    Scale figure size with graph size to reduce label collisions.
    """
    w0, h0 = base
    w = max(w0, min(80.0, 10.0 + 0.45 * n_nodes))
    h = max(h0, min(80.0, 8.0 + 0.30 * n_nodes))
    return (w, h)


def draw_graph(
    G: nx.DiGraph,
    show_edge_labels: bool = False,
    direction: str = "LR",
    node_size: float = 3800,
    max_chars: int = DEFAULT_MAX_CHARS,
    out_png: str = EXPORT_PNG,
    base_figsize: Tuple[float, float] = (22, 16),
) -> None:
    pos = _layout(G, direction=direction, node_size=node_size)

    # Wrapped labels for legibility
    labels = {n: wrap_label(G.nodes[n].get("label", n), max_chars=max_chars) for n in G.nodes}

    figsize = _auto_figsize(len(G.nodes), base=base_figsize)
    plt.figure(figsize=figsize)

    nx.draw_networkx_nodes(
        G,
        pos,
        node_size=node_size,
        node_color=NODE_COLOR,
        edgecolors="black",
        linewidths=1.2,
    )

    nx.draw_networkx_edges(
        G,
        pos,
        arrows=True,
        arrowstyle="-|>",
        arrowsize=ARROWSIZE,
        width=EDGE_WIDTH,
        edge_color=EDGE_COLOR,
        node_size=node_size,
        connectionstyle="arc3,rad=0.02",
    )

    nx.draw_networkx_labels(
        G,
        pos,
        labels=labels,
        font_size=FONT_SIZE,
        font_weight="bold",
    )

    if show_edge_labels:
        edge_labels = {}
        for u, v, data in G.edges(data=True):
            mech = (data.get("mechanism") or "").strip()
            edge_labels[(u, v)] = wrap_label(mech[:80], max_chars=28)
        nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=9)

    plt.axis("off")
    plt.tight_layout()
    plt.savefig(out_png, dpi=500, bbox_inches="tight")
    plt.show()
    print(f"Saved PNG: {out_png}")


def export_graph(G: nx.DiGraph, graphml_path: str, gml_path: str) -> None:
    # GraphML is generally robust for attributes; GML is also commonly used.
    nx.write_graphml(G, graphml_path)
    nx.write_gml(G, gml_path)
    print(f"Exported: {graphml_path}")
    print(f"Exported: {gml_path}")
```

```python
NODES_CSV = "./Papers/nodes_v2.csv"
EDGES_CSV = "./Papers/edges_v2.csv"

if not os.path.exists(NODES_CSV):
    raise FileNotFoundError(f"Missing {NODES_CSV} in current directory.")
if not os.path.exists(EDGES_CSV):
    raise FileNotFoundError(f"Missing {EDGES_CSV} in current directory.")

nodes_df = load_nodes(NODES_CSV)
edges_df = load_edges(EDGES_CSV)

G = build_graph(nodes_df, edges_df)
warn_if_cycles(G)


draw_graph(G, show_edge_labels=False, direction = "LR")

export_graph(G, EXPORT_GRAPHML, EXPORT_GML)

```

```python
def load_json_as_nodes_edges(json_path: str):
    """
    Parse a JSON edge list of the form:
        [source, relation, target]
    into nodes_df and edges_df compatible with build_graph().
    """

    with open(json_path, "r", encoding="utf-8") as f:
        triples = json.load(f)

    # --------------------
    # Build nodes_df
    # --------------------
    nodes = {}
    for src, _, dst in triples:
        src = str(src).strip()
        dst = str(dst).strip()
        if src:
            nodes[src] = {
                "node_id": src,
                "node_name": src,
                "definition_quote": "",
                "parent_node": "",
                "explicitness": "",
                "internal_inconsistency": "",
                "notes": "",
            }
        if dst:
            nodes[dst] = {
                "node_id": dst,
                "node_name": dst,
                "definition_quote": "",
                "parent_node": "",
                "explicitness": "",
                "internal_inconsistency": "",
                "notes": "",
            }

    nodes_df = pd.DataFrame(nodes.values())

    # --------------------
    # Build edges_df
    # --------------------
    edges = []
    for src, rel, dst in triples:
        src = str(src).strip()
        dst = str(dst).strip()
        rel = str(rel).strip()

        if not src or not dst:
            continue

        edges.append({
            "source": src,
            "target": dst,
            "mechanism": rel,
            "temporal_structure": "",
            "context": "",
            "necessity_sufficiency": "",
            "supporting_quote": "",
        })

    edges_df = pd.DataFrame(edges)

    return nodes_df, edges_df
```

```python
def load_json_as_nodes_edges_sean(json_path: str):
    """
    Loads a JSON list of dicts with keys: subject, predicate, object.
    Also repairs a common corruption: missing comma between
        "subject": "..."
        "predicate": "..."
    """

    with open(json_path, "r", encoding="utf-8") as f:
        raw = f.read()

    # Repair: ... "subject": "X"\n    "predicate": ...
    raw = re.sub(r'("subject"\s*:\s*"[^"]+")\s*\n(\s*"predicate")', r"\1,\n\2", raw)

    # If you also ever miss commas before "object", you can uncomment:
    # raw = re.sub(r'("predicate"\s*:\s*"[^"]+")\s*\n(\s*"object")', r"\1,\n\2", raw)

    records = json.loads(raw)

    nodes = {}
    edges = []

    for rec in records:
        subj = str(rec.get("subject", "")).strip()
        pred = str(rec.get("predicate", "")).strip()
        obj  = str(rec.get("object", "")).strip()
        if not subj or not obj:
            continue

        if subj not in nodes:
            nodes[subj] = {
                "node_id": subj, "node_name": subj,
                "definition_quote": "", "parent_node": "",
                "explicitness": "", "internal_inconsistency": "", "notes": ""
            }
        if obj not in nodes:
            nodes[obj] = {
                "node_id": obj, "node_name": obj,
                "definition_quote": "", "parent_node": "",
                "explicitness": "", "internal_inconsistency": "", "notes": ""
            }

        edges.append({
            "source": subj,
            "target": obj,
            "mechanism": pred,
            "temporal_structure": "",
            "context": "",
            "necessity_sufficiency": "",
            "supporting_quote": "",
        })

    return pd.DataFrame(nodes.values()), pd.DataFrame(edges)
```

```python
def load_arrow_txt_as_nodes_edges(txt_path: str):
    """
    Parse a text file with lines of the form:
        NodeA -> NodeB
    into nodes_df and edges_df compatible with build_graph().
    """

    nodes = {}
    edges = []

    with open(txt_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line or "->" not in line:
                continue

            src, dst = [x.strip() for x in line.split("->", 1)]
            if not src or not dst:
                continue

            # ---- nodes ----
            if src not in nodes:
                nodes[src] = {
                    "node_id": src,
                    "node_name": src,
                    "definition_quote": "",
                    "parent_node": "",
                    "explicitness": "",
                    "internal_inconsistency": "",
                    "notes": "",
                }

            if dst not in nodes:
                nodes[dst] = {
                    "node_id": dst,
                    "node_name": dst,
                    "definition_quote": "",
                    "parent_node": "",
                    "explicitness": "",
                    "internal_inconsistency": "",
                    "notes": "",
                }

            # ---- edges ----
            edges.append({
                "source": src,
                "target": dst,
                "mechanism": "causes",
                "temporal_structure": "",
                "context": "",
                "necessity_sufficiency": "",
                "supporting_quote": "",
            })

    nodes_df = pd.DataFrame(nodes.values())
    edges_df = pd.DataFrame(edges)

    return nodes_df, edges_df
```
